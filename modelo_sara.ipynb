```python
# Cargamos el modelo de lenguaje
nlp_es = spacy.load('es_core_news_sm') # Español
nlp_en = spacy.load('en_core_web_sm') # Ingles
columnas = {
    "es": [
        "id_cliente","nombre","apellidos","email","pais","ciudad","edad","genero",
        "id_transaccion","fecha_compra","producto","categoria_producto",
        "precio_unitario","cantidad","importe_total","metodo_pago",
        "coste_envio","coste_fabricacion"
    ],
    "en": [
        "client_id","name","surname","email","country","city","age","gender",
        "transaction_id","purchase_date","product","product_category",
        "unit_price","quantity","total_amount","payment_method",
        "shipping_cost","manufacturing_cost"
    ]
}

```

```python
# Creamos función para detectar el lenguaje y el modelo de lenguaje necesario:
def detectar_idioma(texto):
    # Detectar idioma
    lang = detect(texto)
    if lang == "es":
        return nlp_es(texto), "es"
    elif lang == "en":
        return nlp_en(texto), "en"
    else:
        raise ValueError(f"Idioma no soportado: {lang}")
# Ejemplos
```

```python
detectar_idioma("Muéstrame el promedio de ventas en 2023 por trimestre")
detectar_idioma("Show me the average sales in 2023 by quarter")
def extraer_info(doc, idioma):
    # Entidades reconocidas automáticamente
    entidades = [(ent.text, ent.label_) for ent in doc.ents]
    columnas_detectadas = [token.lemma_ for token in doc if token.lemma_ in columnas[idioma]]
    return entidades, columnas_detectadas
def generar_sql(texto):
    # Primero detectamos idioma y obtener el texto
    doc, idioma = detectar_idioma(texto)
    # Extramos entidades y tokens clave
    entidades, columnas_detectadas = extraer_info(doc, idioma)
    # Contrucción básica de SELECT
    sql = "SELECT "
    if "importe_total" in columnas_detectadas or "total_ammount" in columnas_detectadas:
        sql += "SUM(importe_total) AS total_ingresos"
    elif "cantidad" in columnas_detectadas or "quantity" in columnas_detectadas:
        sql += "SUM(cantidad) AS total_unidades"
    else:
        sql += "*"
    sql += " FROM dataset_merged"
    # Condiciones WHERE según entidades
    condiciones = []
    for ent in entidades:
        if ent[1] == "DATE":
            condiciones.append(f"fecha_compra LIKE '%{ent[0]}%'")
        elif ent[1] == "LOC":
            condiciones.append(f"pais = '{ent[0]}'")
        elif ent[1] == "MONEY":
            condiciones.append(f"importe_total > {ent[0].replace('€','')}")
    if condiciones:
        sql += " WHERE " + " AND ".join(condiciones)
    sql += ";"
    return sql
```

```python
# Ejemplo en español
sql_es = generar_sql("Muéstrame el promedio de ventas en 2023 por trimestre")
print("Consulta SQL (ES):", sql_es)
# Ejemplo en inglés
sql_en = generar_sql("Show me the average sales in 2023 by quarter")
print("Consulta SQL (EN):", sql_en)
## A futuro
#Consolidar con los datos (columnas)
#Generar gráficos
```
