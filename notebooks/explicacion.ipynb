## Imports

```python
import spacy
```

## Cargamos el modelo

```python
nlp_es = spacy.load("es_core_news_sm") # Español
```

## Texto de ejemplo

```python
texto = "Quiero las ventas de móviles de enero del 2023"
```

## Se lo pasamos al modelo

```python
doc = nlp_es(texto)
```

## Segmentación

```python
for s in doc.sents:
    print(s,"\n")
    for token in s:
        print(token.text)
```

## Etiquetado morfológico y lematización (raíz de la palabra)

```python
for token in doc:
    print(token.text, token.lemma_, token.pos_, token.tag_, token.dep_,
            token.shape_, token.is_alpha, token.is_stop)
```

## Análisis sintáctico

```python
spacy.displacy.render(doc, style="dep")
```

## Reconocimiento de entidades

NER (Named Entity Recognition)

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
spacy.displacy.render(doc, style="ent");
```
